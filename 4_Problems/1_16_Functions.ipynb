:## 🟥 Job Posting Summary (1.16.4)

```python
# Code Needed
job_postings = [
    {'title': 'Data Scientist', 'location': 'New York', 'salary': 95000},
    {'title': 'Data Analyst', 'location': 'San Francisco', 'salary': 85000},
    {'title': 'Machine Learning Engineer', 'location': 'New York', 'salary': 115000}
]
```

```python
# Calculate total number of job postings
total_postings = len(job_postings)

# Calculate average salary
total_salaries = sum(post['salary'] for post in job_postings)
average_salary_jobs = total_salaries / total_postings if total_postings > 0 else 0

# Find unique locations and job titles
unique_locations = set(post['location'] for post in job_postings)
unique_titles = set(post['title'] for post in job_postings)

print(f"Total Job Postings: {total_postings}")
print(f"Average Salary: ${average_salary_jobs:,.2f}")
print(f"Unique Locations: {', '.join(unique_locations)}")
print(f"Unique Job Titles: {', '.join(unique_titles)}")
```
